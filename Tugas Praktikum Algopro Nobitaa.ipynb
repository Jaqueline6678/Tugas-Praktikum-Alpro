```python
![image.png](attachment:image.png)
usia_pengemudi = int(input("Masukkan usia pengemudi: "))
jenis_kendaraan = input("Masukkan jenis kendaraan (motor/mobil): ")
nilai_kendaraan = float(input("Masukkan nilai kendaraan (Rp): "))
area_penggunaan = input("Masukkan area penggunaan (perkotaan/pedesaan): ")
sistem_pengaman = input("Apakah kendaraan memiliki sistem pengaman? (ya/tidak): ")

if usia_pengemudi < 20 or usia_pengemudi > 60:
    risiko = 3
elif jenis_kendaraan == "motor":
    risiko = 2
else:
    risiko = 1


if area_penggunaan == "perkotaan":
    risiko = risiko + 1


if sistem_pengaman == "tidak":
    risiko = risiko + 1


if risiko > 3:
    risiko = 3


if risiko == 3:
    kategori = "Tinggi"
    persentase_premi = 0.10
elif risiko == 2:
    kategori = "Sedang"
    persentase_premi = 0.07
else:
    kategori = "Rendah"
    persentase_premi = 0.05


premi = nilai_kendaraan * persentase_premi


pesan = (
    "Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!"
    if kategori == "Tinggi"
    else "Risiko kendaraan masih aman menurut Doraemon."
)

print("\n===== HASIL ANALISIS DORABIKA =====")
print("Kategori risiko:", kategori)
print(f"Premi asuransi: Rp{premi:,.0f}")
print("Pesan Doraemon:", pesan)
```

```python
saldo = 50000
hari = 1

while hari <= 30:
    pengeluaran = int(input(f"Pengeluaran hari ke-{hari}: "))

    saldo -= pengeluaran

    print(f"Sisa saldo hari ke-{hari}: Rp{saldo}")

    if saldo < 0:
        print(f"Saldo habis di hari ke-{hari}. Doraemon menghentikan pencatatan!")
        break

    hari += 1
else:
    print("Pencatatan pengeluaran selama 30 hari selesai.")

```

```python

pengeluaran_mingguan = [20000, 0, 15000, 0]

for i in range(4):
    if pengeluaran_mingguan[i] == 0:
        print(f"Minggu ke-{i+1}: Nobita tidak jajan, dilewati analisisnya.")
        continue

    print(f"Minggu ke-{i+1}: Total pengeluaran Rp{pengeluaran_mingguan[i]}")
else:
    print("Analisis mingguan selesai.")
```

```python

for bulan in range(1, 4):
    print(f"\nBulan ke-{bulan}")
    total_bulanan = 0

    for minggu in range(1, 5):
        pengeluaran = int(input(f"Pengeluaran minggu ke-{minggu}: "))

        if pengeluaran > 20000:
            print("Pengeluaran terlalu besar! Pencatatan bulan ini dihentikan.")
            break

        total_bulanan += pengeluaran

    else:
        print(f"Total pengeluaran bulan ke-{bulan}: Rp{total_bulanan}")

print("\nLaporan bulanan selesai.")
```

```python

saldo_awal = 300000
saldo = saldo_awal

for bulan in range(1, 13):
    print(f"\nBulan ke-{bulan}")
    minggu = 1

    while minggu <= 4:
        libur = input(
            f"Apakah minggu ke-{minggu} libur? (y/n): "
        ).lower()

        if libur == "y":
            print("Minggu libur, dilewati.")
            minggu += 1
            continue

        elif libur == "n":
            pengeluaran = int(
                input(f"Pengeluaran minggu ke-{minggu}: ")
            )
            saldo -= pengeluaran

        else:
            print("Input salah! Masukkan y atau n.")
            continue

        if saldo < 0:
            print("Saldo habis! Doraemon menghentikan evaluasi.")
            break

        minggu += 1

    else:
        print(f"Saldo akhir bulan ke-{bulan}: Rp{saldo}")

    if saldo < 0:
        break

else:
    print(
        f"\nNobita berhasil menabung Rp{saldo} "
        f"dari total saldo awal Rp{saldo_awal}."
    )

```
