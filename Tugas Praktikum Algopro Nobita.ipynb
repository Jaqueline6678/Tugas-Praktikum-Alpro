```python

#Membuat data belanjaan
belanjaan = ["Apel", "Susu", "Roti"]

harga_jumlah = (5000, 3, 12000, 2, 8000, 1)

#Menampilkan data belanjaan
print("Data Belanjaan Nobita")
print("-----------------------------")
print("Barang\tHarga\tJumlah")
print("-----------------------------")
print(f"{belanjaan[0]}\t{harga_jumlah[0]}\t{harga_jumlah[1]}")
print(f"{belanjaan[1]}\t{harga_jumlah[2]}\t{harga_jumlah[3]}")
print(f"{belanjaan[2]}\t{harga_jumlah[4]}\t{harga_jumlah[5]}")

print("-----------------------------")
print("Karakter pertama barang pertama:", belanjaan[0])
print("Karakter terakhir barang terakhir:", belanjaan[-1])
```

```python
#Menghitung total harga belanjaan
total_harga_apel = harga_jumlah[0]*harga_jumlah[1]
total_harga_susu = harga_jumlah[2]*harga_jumlah[3]
total_harga_roti = harga_jumlah[4]*harga_jumlah[5]

total_harga_keseluruhan = total_harga_apel + total_harga_susu + total_harga_roti
print("Total harga apel : Rp", total_harga_apel)
print("Total harga susu : Rp", total_harga_susu)
print("Total harga roti : Rp", total_harga_roti)

print("Total harga belanjaan Nobita : Rp", total_harga_keseluruhan)
```
