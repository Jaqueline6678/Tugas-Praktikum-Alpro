```python
#Data diri Giant
nama = "Giant"
usia = 20               #tahun
berat_badan = "97.5"    #kg
tinggi_badan = 1.75     #meter
status_olahraga = False #bool

berat_badan_float = float(berat_badan)       #ngubah str menjadi float

print("Halo, saya", nama,"usia saya", usia, "tahun dan berat saya", berat_badan_float, "kg")
```

```python
#Menghitung BMI Giant
berat_badan = 97.5 #kg
tinggi_badan = 1.75 #meter

bmi = berat_badan/(tinggi_badan**2)

print("BMI Giant adalah", round(bmi, 2))
```

```python
#Menghitung true division, floor division, dan modulo
target_minum = 2000 #ml air/hari
satu_botol = 350 #ml air

true_division = target_minum/satu_botol
floor_division = target_minum//satu_botol
modulo = target_minum%satu_botol

print("True division:", true_division)
print("Floor division:", floor_division)
print("Modulo:", modulo)
```
